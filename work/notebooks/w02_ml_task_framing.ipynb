## 3. Success metric

*One metric you can defend. What number means 'good'?*


My primary success metric is **Precision@50**.

The practical reason is that an editor has limited time, so the useful
question is how many of the top 50 recommended pages are actually
declining.

A higher Precision@50 means the review queue contains more pages that
deserve attention.

I would also compare the model against a simple hand-written baseline
so that ML has to demonstrate value over a transparent rule.

```python
# Define the success metric for the ranking task.
print("Primary metric: Precision@50")
print("Interpretation: fraction of the top 50 ranked pages that are actually declining.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*


A fixed rule such as "prioritize pages older than 180 days with more
than 500 impressions" is easy to understand, but it uses only a few
hard thresholds.

The starter data contains several observable signals that may interact,
including impressions, average position, CTR, content age, update
recency, and word count.

A model can combine these signals into a ranking without requiring me
to manually choose every threshold and interaction.

ML earns its place only if it improves the ranking quality on unseen
data compared with a simple baseline rule. If a fixed rule performs just
as well, the simpler rule would be preferable.

```python
# Summarize the decision-support role of ML.
print("Baseline: fixed threshold rule")
print("ML advantage to test: combining multiple signals into a ranked priority score")
print("Decision support: prioritize limited editor review time")
```
